Будем использовать учебник поликлинической терапии 2017 года. Материал может содержать потенциальные сложности: большое количество аббревиатур, термины на разных языках, упоминаются разные варианты лечения и разные дозировки препаратов для заболеваний, одинаковые симптомы в контексте разных заболеваний. Загрузим pdf.

```python
loader = PyPDFLoader("/content/pt.pdf")
documents = loader.load()
```

При генерации чанков нужно учесть, что учебник разделен на параграфы, а внутри них - на фрагменты. Используем RecursiveCharacterTextSplitter с сепараторами двойных переносов строк и переносов строк, таким образом в чанк должны попасть объединенные по смыслу данные. Однако часто над новым фрагментом будет какая-то информация с общей информацией или описанием, что будет далее - например, этиология, затем диагностика. Такие 'соседние' данные важны, поэтому установим 20% оверлэп для чанков, пытаясь захватить как можно лучше переходы.

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
docs = splitter.split_documents(documents)
```

Возьмем легкий эмбеддер из HuggingFaceEmbeddings - all-MiniLM-L6-v2

```python
embedder = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

В качестве VectorStore используем Chroma и стандартный ретривер с similarity-поиском.  

```python
db = Chroma.from_documents(docs, embedder)
retriever = db.as_retriever()
```

```python
model = ChatOllama(model="deepseek-r1")
```

Создаем промпт, объединяем контекст из документов в create_stuff_documents_chain, добавляем ретривер.

```python
system_prompt = (
    "Используй ТОЛЬКО предоставленный контекст учебника по поликлинической терапии для ответа на вопрос. "
    "Если ответа нет в контексте, скажи 'Я не знаю'. "
    "Будь кратким - не более 3 предложений. "
    "Контекст: {context}"
)

prompt = ChatPromptTemplate.from_messages([
    ("system", system_prompt),
    ("human", "{input}"),
])

combine_docs_chain = create_stuff_documents_chain(model, prompt)
qa_chain = create_retrieval_chain(retriever, combine_docs_chain)
```

```python
query = "В какой дозировке вводить циклофосфамид?"
response = qa_chain.invoke({"input": query})
answer = response["answer"]
print(answer)
```

Действительно, в тексте указана такая дозировка:

**Применяется циклофосфамид по 2—2,5 мг/кг/сут внутрь в течение
8—12 недель или в виде ежемесячных высокодозных пульсов по 15 мг/кг
внутривенно в течение 6—9 месяцев.**

Однако дозировка 2—2,5 мг/кг/сут не указана в итоговом ответе, хотя упоминается в рассуждениях. Кроме того, скорее всего, курс 15 мг/кг выдается 1 раз в месяц, это кажется логичным, но не указано прямым текстом - 'ежемесячно' можно понять как каждый месяц на протяжении 6-9 месяцев. Однако такой ответ модели можно рассматривать как удовлетворительный - по контексту находится нужная информация, коррекция промпта дала бы более точные ответы модели: вероятно, LLM в поиске подходящего ответа находит самый "эффективный" способ лечения.

```python
query = "Какие методы диагностики у миокардита и перикардита?"
response = qa_chain.invoke({"input": query})
answer = response["answer"]
print(answer)
```

Перикардит упоминается в учебнике в разделах по дифференциальной диагностике, однако четкое описание заболевания отсутствует. Возможно, разделы, где перикардит упомянут вскользь, не распознаются как значимые.

Следующий пример показывает недостатки в работе RAG:

```python
query = "Как диагностировать хроническую болезнь почек (ХБП)"
answer = qa_chain.invoke({"input": query})
answer = response["answer"]
print(answer)
```
